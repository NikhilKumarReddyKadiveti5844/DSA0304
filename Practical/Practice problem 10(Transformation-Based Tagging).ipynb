```python
import nltk

nltk.download("brown")

from nltk.corpus import brown
from nltk.tag import DefaultTagger, brill
from nltk.tag.brill_trainer import BrillTaggerTrainer

train = brown.tagged_sents(categories="news")[:500]

initial_tagger = DefaultTagger("NN")

templates = [
    brill.Template(brill.Pos([-1])),
    brill.Template(brill.Pos([1]))
]

trainer = BrillTaggerTrainer(initial_tagger, templates)
tagger = trainer.train(train, max_rules=10)

text = "The boy is running".split()

result = tagger.tag(text)

for word, tag in result:
    print(word, "->", tag)
```
