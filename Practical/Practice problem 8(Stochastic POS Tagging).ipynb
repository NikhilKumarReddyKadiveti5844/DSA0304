```python
import nltk
import random
from collections import defaultdict, Counter

nltk.download("brown")

from nltk.corpus import brown

words = brown.words()
tags = brown.tagged_words()

model = defaultdict(Counter)

for word, tag in tags:
    model[word.lower()][tag] += 1

text = "The dog runs quickly".split()

for word in text:
    word = word.lower()

    if word in model:
        tag_counts = model[word]
        tag_list = list(tag_counts.keys())
        count_list = list(tag_counts.values())

        tag = random.choices(tag_list, weights=count_list)[0]
    else:
        tag = "NN"

    print(word, "->", tag)
```
